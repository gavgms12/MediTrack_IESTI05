## Inference With The Trained Model

```python
import time
import numpy as np
from PIL import Image
from ultralytics import YOLO
import matplotlib.pyplot as plt
```

```python
model_path= "./models/best.pt"
task = "detect" 
verbose = True
```

```python
model = YOLO(model_path, task, verbose)
```

```python
# Open an image using PIL
source = Image.open("./book-perfume-remedy/image_20250923-202649.jpg")
```

```python
# Run inference on the source
results = model.predict(source, save=False, imgsz=640, conf=0.5, iou=0.3)
result = results[0]
result
```

```python
im_bgr = result.plot()
img = Image.fromarray(im_bgr[..., ::-1])

plt.figure(figsize=(6, 6))
plt.imshow(img)
plt.axis('off')  # This turns off the axis numbers
plt.title("YOLO Result")
plt.show()
```

### Creating a function

```python
def infer_yolo(source, conf=0.5, iou=0.3):
    results = model.predict(source, save=False, imgsz=640, conf=conf, iou=iou)
    result = results[0]

    im_bgr = result.plot()
    img = Image.fromarray(im_bgr[..., ::-1])

    plt.figure(figsize=(6, 6))
    plt.imshow(img)
    plt.axis('off')  # This turns off the axis numbers
    plt.title("YOLO Result")
    plt.show()
```

```python
!ls ./book-perfume-remedy
```

```python
infer_yolo("./book-perfume-remedy/image_20250923-202927.jpg")
```

```python
infer_yolo("./book-perfume-remedy/image_20250923-203131.jpg")
```
